```python
query = """
SELECT Product_Name, Price
FROM Orders
WHERE Price > (
    SELECT AVG(Price)
    FROM Orders
)
ORDER BY Price DESC
"""

df = pd.read_sql(query, con)

plt.figure(figsize=(10,5))
plt.bar(df["Product_Name"], df["Price"])
plt.title("Products Above Average Price")
plt.xlabel("Product")
plt.ylabel("Price")
plt.xticks(rotation=45)
plt.show()
```

```python
query = """
SELECT Customer_Name, Total_Spent
FROM (
    SELECT c.Customer_ID,
           c.Customer_Name,
           SUM(o.Quantity * o.Price) AS Total_Spent
    FROM Customers c
    JOIN Orders o
    ON c.Customer_ID = o.Customer_ID
    GROUP BY c.Customer_ID, c.Customer_Name
) AS customer_sales
WHERE Total_Spent > (
    SELECT AVG(Total_Spent)
    FROM (
        SELECT SUM(Quantity * Price) AS Total_Spent
        FROM Orders
        GROUP BY Customer_ID
    ) AS sales
)
ORDER BY Total_Spent DESC
"""

df = pd.read_sql(query, con)

plt.figure(figsize=(10,5))
plt.bar(df["Customer_Name"], df["Total_Spent"])
plt.title("Customers Spending Above Average")
plt.xlabel("Customer")
plt.ylabel("Total Spent")
plt.xticks(rotation=90)
plt.show()
```

```python
query = """
SELECT Category, Total_Sales
FROM (
    SELECT Category,
           SUM(Quantity * Price) AS Total_Sales
    FROM Orders
    GROUP BY Category
) AS category_sales
WHERE Total_Sales > (
    SELECT AVG(Total_Sales)
    FROM (
        SELECT SUM(Quantity * Price) AS Total_Sales
        FROM Orders
        GROUP BY Category
    ) AS avg_sales
)
"""

df = pd.read_sql(query, con)

plt.figure(figsize=(8,5))
plt.bar(df["Category"], df["Total_Sales"])
plt.title("Categories Above Average Sales")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.show()
```

```python
query = """
SELECT Product_Name, Category, Price
FROM Orders o
WHERE Price > (
    SELECT AVG(Price)
    FROM Orders o2
    WHERE o2.Category = o.Category
)
"""

df = pd.read_sql(query, con)

plt.figure(figsize=(10,5))
plt.bar(df["Product_Name"], df["Price"])
plt.title("Products Above Their Category Average Price")
plt.xlabel("Product")
plt.ylabel("Price")
plt.xticks(rotation=45)
plt.show()
```

```python
query = """
SELECT Product_Name, Quantity
FROM Orders
WHERE Quantity > (
    SELECT AVG(Quantity)
    FROM Orders
)
ORDER BY Quantity DESC
"""

df = pd.read_sql(query, con)

plt.figure(figsize=(10,5))
plt.bar(df["Product_Name"], df["Quantity"])
plt.title("Products With Above Average Quantity")
plt.xlabel("Product")
plt.ylabel("Quantity")
plt.xticks(rotation=45)
plt.show()
```

```python
#joints
#inner joint
query1 = """
SELECT c.Customer_Name,
       SUM(o.Quantity * o.Price) AS Total_Spending
FROM Customers c
INNER JOIN Orders o
ON c.Customer_ID = o.Customer_ID
GROUP BY c.Customer_ID, c.Customer_Name
ORDER BY Total_Spending DESC
"""

df1 = pd.read_sql(query1, con)
df1 = df1.head(10)

plt.figure(figsize=(10,6))
plt.bar(df1["Customer_Name"], df1["Total_Spending"])
plt.title("Customer-wise Total Spending")
plt.xlabel("Customer")
plt.ylabel("Total Spending")
plt.xticks(rotation=90)
plt.show()
```

```python
#left join
query2 = """
SELECT c.Customer_Name,
       COUNT(o.Order_ID) AS Total_Orders
FROM Customers c
LEFT JOIN Orders o
ON c.Customer_ID = o.Customer_ID
GROUP BY c.Customer_ID, c.Customer_Name
ORDER BY Total_Orders DESC
"""

df2 = pd.read_sql(query2, con)
df2 = df2.head(10)


plt.figure(figsize=(10,6))
plt.bar(df2["Customer_Name"], df2["Total_Orders"])
plt.title("Customer-wise Number of Orders")
plt.xlabel("Customer")
plt.ylabel("Number of Orders")
plt.xticks(rotation=90)
plt.show()
```

```python
#right join
query3 = """
SELECT o.Product_Name,
       SUM(o.Quantity * o.Price) AS Total_Sales
FROM Customers c
RIGHT JOIN Orders o
ON c.Customer_ID = o.Customer_ID
GROUP BY o.Product_Name
ORDER BY Total_Sales DESC
"""

df3 = pd.read_sql(query3, con)

plt.figure(figsize=(10,6))
plt.barh(df3["Product_Name"], df3["Total_Sales"])
plt.title("Product-wise Total Sales")
plt.xlabel("Total Sales")
plt.ylabel("Product")
plt.show()
```

```python
query4 = """
SELECT c.City,
       SUM(o.Quantity * o.Price) AS Total_Sales
FROM Customers c
INNER JOIN Orders o
ON c.Customer_ID = o.Customer_ID
GROUP BY c.City
ORDER BY Total_Sales DESC
"""

df4 = pd.read_sql(query4, con)

plt.figure(figsize=(8,5))
plt.bar(df4["City"], df4["Total_Sales"])
plt.title("City-wise Total Sales")
plt.xlabel("City")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.show()
```

```python
query5 = """
SELECT c.Customer_Name,
       c.Age,
       o.Price
FROM Customers c
INNER JOIN Orders o
ON c.Customer_ID = o.Customer_ID
WHERE c.Age >= 25
"""

df5 = pd.read_sql(query5, con)

plt.figure(figsize=(8,5))
plt.scatter(df5["Age"], df5["Price"])
plt.title("Customer Age vs Order Price")
plt.xlabel("Customer Age")
plt.ylabel("Order Price")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Sales by Category
query1 = """
SELECT Category,
       SUM(Quantity * Price) AS Total_Sales
FROM Orders
GROUP BY Category
"""
df1 = pd.read_sql(query1, con)

# 2. Orders by Category
query2 = """
SELECT Category,
       COUNT(*) AS Total_Orders
FROM Orders
GROUP BY Category
"""
df2 = pd.read_sql(query2, con)

# 3. Daily Sales
query3 = """
SELECT Order_Date,
       SUM(Quantity * Price) AS Total_Sales
FROM Orders
GROUP BY Order_Date
ORDER BY Order_Date
"""
df3 = pd.read_sql(query3, con)

# 4. Top 10 Products
query4 = """
SELECT Product_Name,
       SUM(Quantity * Price) AS Total_Sales
FROM Orders
GROUP BY Product_Name
ORDER BY Total_Sales DESC
LIMIT 10
"""
df4 = pd.read_sql(query4, con)


# Dashboard
fig, axes = plt.subplots(2, 2, figsize=(16, 10))

# Chart 1
axes[0,0].bar(df1["Category"], df1["Total_Sales"])
axes[0,0].set_title("Sales by Category")
axes[0,0].set_xlabel("Category")
axes[0,0].set_ylabel("Total Sales")
axes[0,0].tick_params(axis='x', rotation=45)

# Chart 2
axes[0,1].pie(
    df2["Total_Orders"],
    labels=df2["Category"],
    autopct="%1.1f%%"
)
axes[0,1].set_title("Orders by Category")

# Chart 3
axes[1,0].plot(
    df3["Order_Date"],
    df3["Total_Sales"],
    marker="o"
)
axes[1,0].set_title("Daily Sales Trend")
axes[1,0].set_xlabel("Order Date")
axes[1,0].set_ylabel("Sales")
axes[1,0].tick_params(axis='x', rotation=45)

# Chart 4
axes[1,1].barh(
    df4["Product_Name"],
    df4["Total_Sales"]
)
axes[1,1].set_title("Top 10 Products by Sales")
axes[1,1].set_xlabel("Total Sales")

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Sales by Category
query1 = """
SELECT Category,
       SUM(Quantity * Price) AS Total_Sales
FROM Orders
GROUP BY Category
"""
df1 = pd.read_sql(query1, con)


# 2. Orders by Category
query2 = """
SELECT Category,
       COUNT(*) AS Total_Orders
FROM Orders
GROUP BY Category
"""
df2 = pd.read_sql(query2, con)


# 3. Daily Sales
query3 = """
SELECT Order_Date,
       SUM(Quantity * Price) AS Total_Sales
FROM Orders
GROUP BY Order_Date
ORDER BY Order_Date
"""
df3 = pd.read_sql(query3, con)


# 4. Top 10 Products
query4 = """
SELECT Product_Name,
       SUM(Quantity * Price) AS Total_Sales
FROM Orders
GROUP BY Product_Name
ORDER BY Total_Sales DESC
LIMIT 10
"""
df4 = pd.read_sql(query4, con)


# 5. Customer Age vs Spending
query5 = """
SELECT c.Customer_Name,
       c.Age,
       SUM(o.Quantity * o.Price) AS Total_Spending
FROM Customers c
INNER JOIN Orders o
ON c.Customer_ID = o.Customer_ID
GROUP BY c.Customer_ID, c.Customer_Name, c.Age
"""
df5 = pd.read_sql(query5, con)


# 6. Quantity vs Price
query6 = """
SELECT Quantity, Price
FROM Orders
"""
df6 = pd.read_sql(query6, con)


# Create 6-chart dashboard
fig, axes = plt.subplots(2, 3, figsize=(18, 10))


# Chart 1
axes[0,0].bar(df1["Category"], df1["Total_Sales"])
axes[0,0].set_title("Sales by Category")
axes[0,0].set_xlabel("Category")
axes[0,0].set_ylabel("Total Sales")
axes[0,0].tick_params(axis='x', rotation=45)


# Chart 2
axes[0,1].pie(
    df2["Total_Orders"],
    labels=df2["Category"],
    autopct="%1.1f%%"
)
axes[0,1].set_title("Orders by Category")


# Chart 3
axes[0,2].plot(
    df3["Order_Date"],
    df3["Total_Sales"],
    marker="o"
)
axes[0,2].set_title("Daily Sales Trend")
axes[0,2].set_xlabel("Order Date")
axes[0,2].set_ylabel("Sales")
axes[0,2].tick_params(axis='x', rotation=45)


# Chart 4
axes[1,0].barh(
    df4["Product_Name"],
    df4["Total_Sales"]
)
axes[1,0].set_title("Top 10 Products by Sales")
axes[1,0].set_xlabel("Total Sales")


# Chart 5
axes[1,1].scatter(
    df5["Age"],
    df5["Total_Spending"]
)
axes[1,1].set_title("Customer Age vs Spending")
axes[1,1].set_xlabel("Age")
axes[1,1].set_ylabel("Total Spending")


# Chart 6
axes[1,2].scatter(
    df6["Quantity"],
    df6["Price"]
)
axes[1,2].set_title("Quantity vs Price")
axes[1,2].set_xlabel("Quantity")
axes[1,2].set_ylabel("Price")


plt.tight_layout()
plt.show()
```
